### (7) Using the C++ Standard Library, what remains in the stack after these operations? List the contents from bottom to top and identify the top element.

```cpp
std::stack<int> s;
s.push(10);
s.push(5);
if (!s.empty()) s.pop();
s.push(12);
s.push(7);
s.pop();
```

Trace (stack shown from bottom to top):

1. `s.push(10)` → `[10]`
2. `s.push(5)` → `[10, 5]`
3. `s.pop()` → `[10]`
4. `s.push(12)` → `[10, 12]`
5. `s.push(7)` → `[10, 12, 7]`
6. `s.pop()` → `[10, 12]`

The final stack is `[10, 12]`, with `12` at the top.

### (8) Using `std::queue<int>` and `std::stack<int>`, what remains after these operations? List queue contents from front to rear and stack contents from bottom to top.

```cpp
std::queue<int> q;
std::stack<int> s;
q.push(15);
q.push(10);
int x = q.front();
q.pop();
if (!q.empty()) q.pop();
s.push(x);
s.push(2);
s.pop();
s.push(5);
```

Trace (queue shown front-to-rear; stack shown bottom-to-top):

1. `q.push(15)` → queue `[15]`
2. `q.push(10)` → queue `[15, 10]`
3. `x = q.front(); q.pop()` → `x = 15`, queue `[10]`
4. `q.pop()` → queue `[]`
5. `s.push(x)` → stack `[15]`
6. `s.push(2)` → stack `[15, 2]`
7. `s.pop()` → stack `[15]`
8. `s.push(5)` → stack `[15, 5]`

The queue is empty. The final stack is `[15, 5]`, with `5` at the top.
